```python
# # Matplotlib Assignment2

# BAR CHART :
print("** BAR CHART :: **")

# Q1.Create a Vertical Bar Chart showing the sales of four products: 
# ● Laptop 
# ● Mobile 
# ● Tablet 
# ● Watch 

import matplotlib.pyplot as plt

products = ["Laptop", "Mobile", "Tablet", "Watch"]
sales = [50, 80, 40, 60]

print("QUESTION 1")
plt.bar(products, sales)
plt.title("Product Sales")
plt.xlabel("Products")
plt.ylabel("Sales")
plt.show()
print()

# Q2.Create a Horizontal Bar Chart showing employee salaries of five employees. 

employees = ["A", "B", "C", "D", "E"]
salary = [25000, 32000, 40000, 38000, 45000]

print("QUESTION 2")
plt.barh(employees, salary)
plt.title("Employee Salaries")
plt.xlabel("Salary")
plt.ylabel("Employees")
plt.show()
print()
    
# Q3.Create a Bar Chart showing the marks of six students. 
# Requirements: 
# ● Green bars 
# ● Black border 
# ● Grid 

students = ["S1","S2","S3","S4","S5","S6"]
marks = [75,80,68,90,85,78]

print("QUESTION 3")
plt.bar(students, marks, color="green", edgecolor="black")
plt.grid()
plt.show()
print()

# Q4.Create a Product Sales Bar Chart using different colors for each bar. 

products = ["Laptop","Mobile","Tablet","Watch"]
sales = [50,80,40,60]

colors = ["red","blue","green","orange"]

print("QUESTION 4")
plt.bar(products, sales, color=colors)
plt.show()
print()

# Q5.Create a Bar Chart showing monthly expenses. 
# Requirements: 
# ● Figure Size (8,5) 
# ● Orange bars 
# ● Horizontal Grid 

months = ["Jan","Feb","Mar","Apr","May"]
expenses = [12000,15000,14000,16000,18000]

print("QUESTION 5")
plt.figure(figsize=(8,5))

plt.bar(months, expenses, color="orange")
plt.grid(axis="y")
plt.show()
print()

# PIE CHART :
print("** PIE CHART :: **")

# Q6.Create a Pie Chart showing the percentage of payment methods: 
# ● Cash 
# ● UPI 
# ● Card 
# ● Net Banking 

methods = ["Cash","UPI","Card","Net Banking"]
values = [30,40,20,10]

print("QUESTION 6")

plt.pie(values, labels=methods)
plt.title("Payment Methods")
plt.show()
print()
    
# Q7.Create a Pie Chart for monthly expenses. 
# Requirements: 
# ● Labels 
# ● Percentages (autopct) 
# ● Shadow 

labels = ["Food","Rent","Travel","Shopping"]
expense = [30,40,20,10]

print("QUESTION 7")
plt.pie(expense, labels=labels, autopct="%1.1f%%", shadow=True)
plt.title("Monthly Expenses")
plt.show()
print()

# Q8.Create a Pie Chart showing product-wise sales. 
# Requirements: 
# ● Different colors 
# ● Start Angle 90 
# ● Title 

products = ["Laptop","Mobile","Tablet","Watch"]
sales = [40,30,20,10]

colors = ["red","blue","green","orange"]

print("QUESTION 8")
plt.pie(sales, labels=products, colors=colors, startangle=90)
plt.title("Product Sales")
plt.show()    
print()

# Q9.Create a Pie Chart showing city-wise population. 
# Requirements: 
# ● Explode the largest slice 
# ● Percentages 
# ● Shadow 

cities = ["Pune","Mumbai","Nashik","Nagpur"]
population = [50,80,30,20]

explode = [0,0.1,0,0]

print("QUESTION 9")
plt.pie(population, labels=cities, explode=explode, autopct="%1.1f%%", shadow=True)
plt.title("City Population")
plt.show()
print()

# Q10.Create a Pie Chart showing department-wise employees. 
# Requirements: 
# ● Labels 
# ● Colors 
# ● Percentages 
# ● Shadow 
# ● Start Angle 90 

departments = ["HR","IT","Sales","Finance"]
employees = [20,35,25,20]

colors = ["red","green","blue","orange"]

print("QUESTION 10")
plt.pie(employees, labels=departments, colors=colors, autopct="%1.1f%%", shadow=True, startangle=90)
plt.title("Department Employees")
plt.show()
print()

# HISTOGRAM :
print("** HISTOGRAM :: **")

# Q11.Create a Histogram showing student marks. 
# Requirements: 
# ● 5 bins 
# ● Green color 
# ● Black border 

marks = [55,60,65,70,72,75,80,82,85,90]

print("QUESTION 11")
plt.hist(marks, bins=5, color="green", edgecolor="black")
plt.show()
print()

# Q12.Create a Histogram showing employee salaries. 
# Requirements: 
# ● 6 bins 
# ● Orange color 
# ● Grid 

salary = [25000,27000,30000,32000,35000,37000,40000,42000,45000]

print("QUESTION 12")
plt.hist(salary, bins=6, color="orange")
plt.grid()
plt.show()
print()
    
# Q13.Create a Histogram showing the age distribution of 25 people. 

ages = [18,20,22,25,30,35,28,24,21,23,
        27,29,32,26,31,34,36,38,40,19,
        22,24,26,28,30]

print("QUESTION 13")
plt.hist(ages)
plt.title("Age Distribution")
plt.show()
print()

# Q14.Create a Histogram showing customer ratings. 
# Requirements: 
# ● Blue color 
# ● 8 bins 
# ● Black border 

ratings = [1,2,2,3,3,4,4,5,5,5,4,3,2,5,4]

print("QUESTION 14")
plt.hist(ratings, bins=8, color="blue", edgecolor="black")
plt.show()
print()

# Q15.Create a Histogram showing daily temperatures for one month. 
# Requirements: 
# ● 7 bins 
# ● Title 
# ● Labels 
# ● Grid 

temperature = [30,31,29,32,34,35,36,33,32,31,
               30,29,28,27,31,32,34,35,33,30,
               29,31,32,34,35,36,33,32,31,30]

print("QUESTION 15")
plt.hist(temperature, bins=7)
plt.title("Daily Temperature")
plt.xlabel("Temperature")
plt.ylabel("Frequency")
plt.grid()
plt.show()
print()

# SCATTER PLOT : 
print("**SCATTER PLOT :: **") 

# Q16.Create a Scatter Plot showing Study Hours vs Marks. 
# Requirements: 
# ● Red color 
# ● Circle marker 
# ● Marker size 100 

hours = [1,2,3,4,5]
marks = [50,60,70,85,95]

print("QUESTION 16")
plt.scatter(hours, marks, color="red", marker="o", s=100)
plt.show()
print()

# Q17.Create a Scatter Plot showing Height vs Weight. 
# Requirements: 
# ● Blue color 
# ● Diamond marker 
# ● Grid 

height = [150,155,160,165,170]
weight = [45,50,55,60,68]

print("QUESTION 17")
plt.scatter(height, weight, color="blue", marker="D")
plt.grid()
plt.show()
print()

# Q18.Create a Scatter Plot showing Experience vs Salary. 
# Requirements: 
# ● Green color 
# ● Marker size 150 
# ● alpha = 0.7 

experience = [1,2,3,4,5]
salary = [25000,30000,35000,42000,50000]

print("QUESTION 18")
plt.scatter(experience, salary, color="green", s=150, alpha=0.7)
plt.show()
print()

# Q19.Create a Scatter Plot showing Quantity vs Total Sales. 
# Requirements: 
# ● Purple color 
# ● Star (*) marker 
# ● Grid 

quantity = [10,20,30,40,50]
sales = [1000,2500,3500,5000,6500]

print("QUESTION 19")
plt.scatter(quantity, sales, color="purple", marker="*")
plt.grid()
plt.show()
print()

# Q20.Create a Scatter Plot showing Age vs Income. 
# Requirements: 
# ● Orange color 
# ● Square (s) marker 
# ● Marker size 120 

age = [20,25,30,35,40]
income = [20000,30000,40000,50000,60000]

print("QUESTION 20")
plt.scatter(age, income, color="orange", marker="s", s=120)
plt.show()
print()

# CHART CUSTOMIZATION : 
print("** CHART CUSTOMIZATION :: **")

# Q21.Create a Product Sales Bar Chart with: 
# ● Figure Size (10,5) 
# ● Green bars 
# ● Black border 
# ● Title 
# ● Labels 
# ● Horizontal Grid 

products = ["Laptop","Mobile","Tablet","Watch"]
sales = [50,80,40,60]

print("QUESTION 21")

plt.figure(figsize=(10,5))
plt.bar(products, sales, color="green", edgecolor="black")

plt.title("Product Sales")
plt.xlabel("Products")
plt.ylabel("Sales")
plt.grid(axis="y")
plt.show()
print()

# Q22.Create a Student Marks Histogram with: 
# ● 5 bins 
# ● Purple color 
# ● Black border 
# ● Title 
# ● Labels 
# ● Grid 

marks = [55,60,65,70,72,75,80,82,85,90]

print("QUESTION 22")

plt.hist(marks, bins=5, color="purple", edgecolor="black")

plt.title("Student Marks")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.grid()
plt.show()
print()

# Q23.Create a Height vs Weight Scatter Plot with: 
# ● Diamond marker 
# ● Marker size 150 
# ● Blue color 
# ● alpha = 0.8 
# ● Title 
# ● Labels 
# ● Grid 

height = [150,155,160,165,170]
weight = [45,50,55,60,68]

print("QUESTION 23")

plt.scatter(height, weight, marker="D", s=150, color="blue", alpha=0.8)

plt.title("Height vs Weight")
plt.xlabel("Height")
plt.ylabel("Weight")
plt.grid()
plt.show()
print()

# Q24.Create a Product Sales Pie Chart with: 
# ● Labels 
# ● Percentages 
# ● Shadow 
# ● Explode the highest sales category 
# ● Start Angle 90 
# ● Title 

products = ["Laptop","Mobile","Tablet","Watch"]
sales = [50,80,40,60]

explode = [0,0.1,0,0]

print("QUESTION 24")

plt.pie(sales, labels=products, autopct="%1.1f%%", shadow=True, explode=explode, startangle=90)

plt.title("Product Sales")
plt.show()
print()

# Q25. Final Practical Challenge 
# Create a Sales Analysis Dashboard containing: 
# ●      Product Sales (Bar Chart) 
# ●       Payment Method (Pie Chart) 
# ●       Student Marks Distribution (Histogram) 
# ●       Study Hours vs Marks (Scatter Plot) 
# Requirements: 
# ● Use plt.subplot() to display all 4 charts in one figure. 
# ● Add a main title using plt.suptitle(). 
# ● Add titles to each chart. 
# ● Add labels wherever applicable. 
# ● Add grid to Bar Chart, Histogram, and Scatter Plot. 
# ● Set figure size to (12, 8).

print("QUESTION 25")

plt.figure(figsize=(12,8))

# 1. Bar Chart
plt.subplot(2,2,1)
products = ["Laptop","Mobile","Tablet","Watch"]
sales = [50,80,40,60]
plt.bar(products, sales, color="green")
plt.title("Product Sales")
plt.xlabel("Products")
plt.ylabel("Sales")
plt.grid()

# 2. Pie Chart
plt.subplot(2,2,2)
methods = ["Cash","UPI","Card","Net Banking"]
payment = [30,40,20,10]
plt.pie(payment, labels=methods, autopct="%1.1f%%", shadow=True)
plt.title("Payment Method")

# 3. Histogram
plt.subplot(2,2,3)
marks = [55,60,65,70,72,75,80,82,85,90]
plt.hist(marks, bins=5, color="purple", edgecolor="black")
plt.title("Student Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.grid()

# 4. Scatter Plot
plt.subplot(2,2,4)
hours = [1,2,3,4,5]
marks = [50,60,70,85,95]
plt.scatter(hours, marks, color="red", s=100)
plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.grid()

plt.suptitle("Sales Analysis Dashboard")
plt.tight_layout()
plt.show()
```
