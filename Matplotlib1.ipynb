```python
## Matplotlib – Day 1 Assignment 
## Topic: Introduction to Matplotlib (Line Charts)

# Question 1: Create your first line chart using the following data: 
# ● X-axis: [1, 2, 3, 4, 5] 
# ● Y-axis: [10, 20, 15, 25, 30] 
# Display the chart. 

import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 25, 30]
print("QUESTION 1")
plt.plot(x, y)
plt.show()
print()

# Question 2: Create a student marks line chart for five subjects. 
# Marks: 
# [72, 80, 76, 90, 95] 
# Add: 
# ● Chart Title 
# ● X-axis Label 
# ● Y-axis Label 

subjects = ["Math", "Science", "English", "Computer", "History"]
marks = [72, 80, 76, 90, 95]

print("QUESTION 2")
plt.plot(subjects, marks)

plt.title("Student Marks")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.show()
print()

# Question 3: Create an employee salary line chart using the following data: 
# [25000, 32000, 40000, 38000, 45000] 
# Requirements: 
# ● Blue Line 
# ● Circle (o) Marker 
# ● Grid 

employees = ["Emp1", "Emp2", "Emp3", "Emp4", "Emp5"]
salary = [25000, 32000, 40000, 38000, 45000]

print("QUESTION 3")
plt.plot(employees, salary, color="blue", marker="o")
plt.title("Employee Salary")
plt.xlabel("Employees")
plt.ylabel("Salary (₹)")
plt.grid()
plt.show()
print()

# Question 4: Plot the monthly sales for six months. 
# Sales: 
# [25000, 30000, 35000, 40000, 45000, 50000] 
# Requirements: 
# ● Green Line 
# ● Dashed (--) Line Style 
# ● Line Width = 3 

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [25000, 30000, 35000, 40000, 45000, 50000]

print("QUESTION 4")
plt.plot(months, sales, color="green", linestyle="--", linewidth=3)
plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales (₹)")
plt.show() 
print()
             
# Question 5: Create a line chart for the following temperature data: 
# [30, 32, 31, 35, 36, 34, 33] 
# Requirements: 
# ● Star (*) Marker 
# ● Red Line 
# ● Chart Title 

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
temperature = [30, 32, 31, 35, 36, 34, 33]

print("QUESTION 5")
plt.plot(days, temperature, color="red", marker="*")
plt.title("Weekly Temperature")
plt.xlabel("Days")
plt.ylabel("Temperature (°C)")
plt.show()
print()

# Question 6: Compare the marks of Student A and Student B on the same graph. 
# Student A: 
# [75, 80, 78, 90, 88] 
# Student B: 
# [70, 85, 80, 86, 92] 
# Requirements: 
# ● Different Colors 
# ● Legend 
# ● Grid 

subjects = ["Math", "Science", "English", "Computer", "History"]

student_A = [75, 80, 78, 90, 88]
student_B = [70, 85, 80, 86, 92]

print("QUESTION 6")

plt.plot(subjects, student_A, color="blue", marker="o", label="Student A")

plt.plot(subjects, student_B, color="red", marker="o", label="Student B")

plt.title("Student Marks Comparison")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.legend()
plt.grid()
plt.show()    
print()

# Question 7: Create an attendance chart for five days. 
# Attendance: 
# [40, 42, 39, 45, 44] 
# Requirements: 
# ● Figure Size (8,5) 
# ● Purple Line 
# ● Diamond (D) Marker 
# ● X-axis Label 
# ● Y-axis Label 

days = ["Mon", "Tue", "Wed", "Thu", "Fri"]
attendance = [40, 42, 39, 45, 44]

print("QUESTION 7")

plt.figure(figsize=(8,5))

plt.plot(days, attendance, color="purple", marker="D")
plt.title("Student Attendance")
plt.xlabel("Days")
plt.ylabel("Attendance")
plt.show()    
print()

# Question 8: Create a line chart showing product sales. 
# Sales Data: 
# [120, 180, 160, 220, 250] 
# Requirements: 
# ● Orange Line 
# ● Dotted (:) Line Style 
# ● Circle (o) Marker 
# ● Grid 

products = ["Laptop", "Mobile", "Tablet", "Watch", "Camera"]
sales = [120, 180, 160, 220, 250]

print("QUESTION 8")
plt.plot(products, sales, color="orange", linestyle=":", marker="o")
plt.title("Product Sales")
plt.xlabel("Products")
plt.ylabel("Sales")
plt.grid()
plt.show()    
print()

# Question 9: Create a line chart for website visitors over seven days. 
# Visitors: 
# [150, 180, 170, 200, 230, 210, 250] 
# Requirements: 
# ● Figure Size (10,5) 
# ● Black Line 
# ● Plus (+) Marker 
# ● Title 
# ● Legend 
# ● Grid 

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
visitors = [150, 180, 170, 200, 230, 210, 250]

print("QUESTION 9")
plt.figure(figsize=(10,5))

plt.plot(days, visitors, color="black", marker="+", label="Visitors")
plt.title("Website Visitors")
plt.xlabel("Days")
plt.ylabel("Number of Visitors")
plt.legend()
plt.grid()
plt.show()   
print()

# Question 10: Create a professional Monthly Sales Report. 
# Months: 
# ["Jan", "Feb", "Mar", "Apr", "May", "Jun"] 
# Sales: 
# [25000, 30000, 28000, 40000, 45000, 50000] 
# Requirements: 
# ● Figure Size (10,5) 
# ● Green Line 
# ● Circle (o) Marker 
# ● Dashed (--) Line Style 
# ● Line Width = 3 
# ● Chart Title 
# ● X-axis Label 
# ● Y-axis Label 
# ● Grid 
# ● Legend

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [25000, 30000, 28000, 40000, 45000, 50000]

print("QUESTION 10")
plt.figure(figsize=(10,5))

plt.plot(months, sales, color="green", marker="o", linestyle="--", linewidth=3, label="Monthly Sales")

plt.title("Monthly Sales Report")
plt.xlabel("Months")
plt.ylabel("Sales (₹)")
plt.grid()
plt.legend()
plt.show()
```
