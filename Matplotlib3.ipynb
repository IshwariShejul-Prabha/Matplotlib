```python
## Matplotlib - Assignment3

# Part A – subplot() (Questions 1–6) 
print("Part A – subplot() ")

# Q1.Create a figure containing 2 subplots: 
# ● Line Chart 
# ● Bar Chart 

import matplotlib.pyplot as plt

print("QUESTION 1")
plt.figure(figsize=(10,4))

# Line Chart
plt.subplot(1,2,1)
x = [1,2,3,4,5]
y = [10,20,15,25,30]
plt.plot(x, y, marker="o", color="blue")
plt.title("Line Chart")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid()

# Bar Chart
plt.subplot(1,2,2)
products = ["Laptop","Mobile","Tablet","Watch"]
sales = [50,80,40,60]
plt.bar(products, sales, color="green")
plt.title("Bar Chart")
plt.xlabel("Products")
plt.ylabel("Sales")
plt.grid(axis="y")

plt.tight_layout()
plt.show()
print()


# Q2.Create a figure containing 4 subplots: 
# ● Line Chart 
# ● Bar Chart 
# ● Pie Chart 
# ● Scatter Plot 

print("QUESTION 2")
plt.figure(figsize=(10,8))

# Line Chart
plt.subplot(2,2,1)
plt.plot([1,2,3,4],[10,20,15,25], marker="o")
plt.title("Line Chart")
plt.grid()

# Bar Chart
plt.subplot(2,2,2)
plt.bar(["A","B","C","D"], [20,35,25,30], color="orange")
plt.title("Bar Chart")
plt.grid(axis="y")

# Pie Chart
plt.subplot(2,2,3)
plt.pie([30,40,20,10],
        labels=["Cash","UPI","Card","Net Banking"],
        autopct="%1.1f%%")
plt.title("Pie Chart")

# Scatter Plot
plt.subplot(2,2,4)
plt.scatter([1,2,3,4,5],
            [50,60,70,85,95],
            color="red")
plt.title("Scatter Plot")
plt.grid()

plt.tight_layout()
plt.show()
print()


# Q3.Create a 2 × 2 Student Dashboard showing: 
# ● Student Marks 
# ● Attendance 
# ● Subject Distribution 
# ● Study Hours vs Marks 

print("QUESTION 3")
plt.figure(figsize=(10,8))

# Student Marks
plt.subplot(2,2,1)
subjects = ["Math","Science","English","Computer","History"]
marks = [72,80,76,90,95]
plt.plot(subjects, marks, marker="o")
plt.title("Student Marks")
plt.grid()

# Attendance
plt.subplot(2,2,2)
days = ["Mon","Tue","Wed","Thu","Fri"]
attendance = [40,42,39,45,44]
plt.bar(days, attendance, color="green")
plt.title("Attendance")
plt.grid(axis="y")

# Subject Distribution
plt.subplot(2,2,3)
plt.pie([20,25,15,30,10],
        labels=subjects,
        autopct="%1.1f%%")
plt.title("Subject Distribution")

# Study Hours vs Marks
plt.subplot(2,2,4)
hours = [1,2,3,4,5]
marks = [55,65,75,88,95]
plt.scatter(hours, marks, color="purple")
plt.title("Study Hours vs Marks")
plt.xlabel("Hours")
plt.ylabel("Marks")
plt.grid()

plt.tight_layout()
plt.show()
print()


# Q4.Create an Employee Dashboard containing: 
# ● Salary Trend 
# ● Department-wise Employees 
# ● Salary Distribution (Histogram) 
# ● Experience vs Salary 

print("QUESTION 4")
plt.figure(figsize=(10,8))

# Salary Trend
plt.subplot(2,2,1)
salary = [25000,32000,40000,38000,45000]
plt.plot(salary, marker="o", color="blue")
plt.title("Salary Trend")
plt.grid()

# Department-wise Employees
plt.subplot(2,2,2)
departments = ["HR","IT","Sales","Finance"]
employees = [20,35,25,15]
plt.bar(departments, employees, color="orange")
plt.title("Department-wise Employees")
plt.grid(axis="y")

# Salary Distribution
plt.subplot(2,2,3)
plt.hist(salary, bins=5, color="green", edgecolor="black")
plt.title("Salary Distribution")
plt.grid()

# Experience vs Salary
plt.subplot(2,2,4)
experience = [1,2,3,4,5]
plt.scatter(experience, salary, color="red")
plt.title("Experience vs Salary")
plt.xlabel("Experience")
plt.ylabel("Salary")
plt.grid()

plt.tight_layout()
plt.show()
print()


# Q5.Create a dashboard with: 
# ● Monthly Sales (Line) 
# ● Product Sales (Bar) 
# ● Payment Methods (Pie) 
# ● Customer Ratings (Histogram) 
# Use subplot() to display all charts together. 

print("QUESTION 5")
plt.figure(figsize=(10,8))

# Monthly Sales
plt.subplot(2,2,1)
months = ["Jan","Feb","Mar","Apr","May","Jun"]
sales = [25000,30000,35000,40000,45000,50000]
plt.plot(months, sales, marker="o")
plt.title("Monthly Sales")
plt.grid()

# Product Sales
plt.subplot(2,2,2)
products = ["Laptop","Mobile","Tablet","Watch"]
product_sales = [50,80,40,60]
plt.bar(products, product_sales, color="green")
plt.title("Product Sales")
plt.grid(axis="y")

# Payment Methods
plt.subplot(2,2,3)
payment = [30,40,20,10]
labels = ["Cash","UPI","Card","Net Banking"]
plt.pie(payment, labels=labels, autopct="%1.1f%%")
plt.title("Payment Methods")

# Customer Ratings
plt.subplot(2,2,4)
ratings = [2,3,4,5,4,3,5,4,5,3,4]
plt.hist(ratings, bins=5, color="orange", edgecolor="black")
plt.title("Customer Ratings")
plt.grid()

plt.tight_layout()
plt.show()
print()


# Q6.Create a 3 × 2 subplot layout and display any six different charts learned in Matplotlib. 

print("QUESTION 6")
plt.figure(figsize=(12,10))

# Line Chart
plt.subplot(3,2,1)
plt.plot([1,2,3,4],[10,20,15,25], marker="o")
plt.title("Line Chart")
plt.grid()

# Bar Chart
plt.subplot(3,2,2)
plt.bar(["A","B","C"], [20,35,30], color="green")
plt.title("Bar Chart")
plt.grid(axis="y")

# Pie Chart
plt.subplot(3,2,3)
plt.pie([40,30,20,10],
        labels=["A","B","C","D"],
        autopct="%1.1f%%")
plt.title("Pie Chart")

# Histogram
plt.subplot(3,2,4)
plt.hist([10,20,20,30,40,30,25,20],
         bins=5,
         color="purple",
         edgecolor="black")
plt.title("Histogram")
plt.grid()

# Scatter Plot
plt.subplot(3,2,5)
plt.scatter([1,2,3,4,5],
            [15,25,35,30,40],
            color="red")
plt.title("Scatter Plot")
plt.grid()

# Horizontal Bar Chart
plt.subplot(3,2,6)
plt.barh(["X","Y","Z"], [10,20,15], color="orange")
plt.title("Horizontal Bar Chart")
plt.grid(axis="x")

plt.tight_layout()
plt.show()
print()
```

```python
# Q25. Final Practical Challenge – Supermarket Sales Dashboard 
# Using a single CSV dataset, create a complete dashboard. 
# Requirements 
# Read the CSV file using Pandas. 
# Create: 
# ●      Monthly Sales (Line Chart) 
# ●      Product Sales (Bar Chart) 
# ●      Payment Method Distribution (Pie Chart) 
# ●      Customer Rating Distribution (Histogram) 
# ●      Quantity vs Total Amount (Scatter Plot) 
# Additional Requirements 
# ● Use subplot() to display all charts in one figure. 
# ● Set figure size to (15,10). 
# ● Add individual titles for each chart. 
# ● Add a main title using plt.suptitle(). 
# ● Use ggplot style. 
# ● Apply tight_layout(). 
# ● Save the dashboard as: 
# sales_dashboard.png 
# using: 
# ● dpi = 300

print("QUESTION 25")

import pandas as pd
import matplotlib.pyplot as plt

# Read CSV File
df = pd.read_csv("sales.csv")

# Apply Style
plt.style.use("ggplot")

# Figure Size
plt.figure(figsize=(15,10))

# Monthly Sales
plt.subplot(2,3,1)
plt.plot(df["Month"], df["Sales"], marker="o")

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid()

# Product Sales
plt.subplot(2,3,2)
plt.bar(df["Product"], df["Sales"], color="green")

plt.title("Product Sales")
plt.xlabel("Product")
plt.ylabel("Sales")
plt.grid(axis="y")

# Payment Method Distribution
plt.subplot(2,3,3)
payment = df["Payment Method"].value_counts()

plt.pie(payment, labels=payment.index, autopct="%1.1f%%", startangle=90)
plt.title("Payment Method")

# Customer Rating Distribution
plt.subplot(2,3,4)
plt.hist(df["Rating"], bins=5, color="orange", edgecolor="black")

plt.title("Customer Ratings")
plt.xlabel("Rating")
plt.ylabel("Frequency")
plt.grid()

# Quantity vs Total Amount
plt.subplot(2,3,5)
plt.scatter(df["Quantity"], df["Total Amount"], color="red")

plt.title("Quantity vs Total Amount")
plt.xlabel("Quantity")
plt.ylabel("Total Amount")
plt.grid()

plt.suptitle("Supermarket Sales Dashboard")
plt.tight_layout()

plt.savefig("sales_dashboard.png", dpi=300)
plt.show()
print()
```
